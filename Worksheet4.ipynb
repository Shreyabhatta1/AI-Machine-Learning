```python
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder

# Example: Create dummy data (replace this with MNIST loading if needed)
X = np.random.rand(1000, 28, 28)   # 1000 images (like MNIST)
y = np.random.randint(0, 10, 1000) # labels 0–9

# Split data
x_train, x_test, y_train_raw, y_test = train_test_split(X, y, test_size=0.2)

# Flatten images
x_train_flat = x_train.reshape(x_train.shape[0], -1)

# One-hot encode labels
encoder = OneHotEncoder(sparse_output=False)
y_train = encoder.fit_transform(y_train_raw.reshape(-1, 1))

# Initialize weights
weights = np.random.rand(x_train_flat.shape[1], y_train.shape[1])

learning_rate = 0.01

# Training loop
for epoch in range(10):
    # Forward pass
    y_pred = np.dot(x_train_flat, weights)

    # Loss (MSE)
    loss = np.mean((y_pred - y_train) ** 2)

    # Gradients
    gradients = 2 * np.dot(x_train_flat.T, (y_pred - y_train)) / len(x_train_flat)

    # Update weights
    weights -= learning_rate * gradients

    print(f"Epoch {epoch}, Loss: {loss}")
```

```python
import os
import numpy as np
import tensorflow as tf
from tensorflow.keras.utils import to_categorical
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt
from PIL import Image, UnidentifiedImageError  # Import Pillow

# Define dataset paths
train_dir = "/content/drive/MyDrive/AI ML/DevanagariHandwrittenDigitDataset/Train"
test_dir = "/content/drive/MyDrive/AI ML/DevanagariHandwrittenDigitDataset/Test"

# Define image size
img_height, img_width = 28, 28

# Function to load images and labels using PIL
def load_images_from_folder(folder):
    images = []
    labels = []
    class_names = sorted(os.listdir(folder))
    class_map = {name: i for i, name in enumerate(class_names)}

    for class_name in class_names:
        class_path = os.path.join(folder, class_name)
        if not os.path.isdir(class_path):
            continue  # skip if not a folder
        label = class_map[class_name]

        for filename in os.listdir(class_path):
            img_path = os.path.join(class_path, filename)

            # Skip non-image files
            if not filename.lower().endswith(('.png', '.jpg', '.jpeg')):
                continue

            try:
                img = Image.open(img_path).convert("L")
                img = img.resize((img_width, img_height))
                img = np.array(img) / 255.0
                images.append(img)
                labels.append(label)
            except UnidentifiedImageError:
                print(f"Skipping bad image: {img_path}")
                continue

    return np.array(images), np.array(labels)

# Load training and testing datasets
x_train, y_train = load_images_from_folder(train_dir)
x_test, y_test = load_images_from_folder(test_dir)

# Reshape images for Keras input
x_train = x_train.reshape(-1, img_height, img_width, 1)
x_test = x_test.reshape(-1, img_height, img_width, 1)

# One-hot encode labels
y_train = to_categorical(y_train, num_classes=10)
y_test = to_categorical(y_test, num_classes=10)

# Print dataset shape
print(f"Training set: {x_train.shape}, Labels: {y_train.shape}")
print(f"Testing set: {x_test.shape}, Labels: {y_test.shape}")

# Visualize some images
plt.figure(figsize=(10, 4))
for i in range(10):
    plt.subplot(2, 5, i + 1)
    plt.imshow(x_train[i].reshape(28, 28), cmap='gray')
    plt.title(f"Label: {np.argmax(y_train[i])}")
    plt.axis("off")
plt.show()
```

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.datasets import mnist
# Load the MNIST dataset
(x_train, y_train), (x_test, y_test) = mnist.load_data()
# Normalize the images to values between 0 and 1
x_train, x_test = x_train / 255.0, x_test / 255.0
# Flatten the 28x28 images into 784-dimensional vectors
x_train = x_train.reshape(-1, 28 * 28)
x_test = x_test.reshape(-1, 28 * 28)
# One-hot encode the labels (0-9) for classification
y_train = tf.keras.utils.to_categorical(y_train, 10)
y_test = tf.keras.utils.to_categorical(y_test, 10)
```

```python
# Model parameters
import tensorflow as tf
from tensorflow import keras
num_classes = 10
input_shape = (28, 28, 1)
model = keras.Sequential(
   [
       keras.layers.Input(shape=input_shape),
       keras.layers.Flatten(), # Flatten the 28x28 image to a 784-dimensional vector
       keras.layers.Dense(64, activation="sigmoid"),
       keras.layers.Dense(128, activation="sigmoid"),
       keras.layers.Dense(256, activation="sigmoid"),
       keras.layers.Dense(num_classes, activation="softmax"),
]
)
```

```python
# Model parameters
import tensorflow as tf
from tensorflow import keras
num_classes = 10
input_shape = (28, 28, 1)
def build_functional_model():
  # Input layer
    inputs = keras.Input(shape=input_shape)
    # Flatten layer
    x = keras.layers.Flatten()(inputs)
    # Hidden layers
    x = keras.layers.Dense(64, activation="sigmoid")(x)
    x = keras.layers.Dense(128, activation="sigmoid")(x)
    x = keras.layers.Dense(256, activation="sigmoid")(x)
    # Output layer
    outputs = keras.layers.Dense(num_classes, activation="softmax")(x)
    # Create model
    model = keras.Model(inputs=inputs, outputs=outputs)
    return model
#Build the model
functional_model = build_functional_model()
functional_model.summary()

```

Task1: Data Preparation

```python
import os
import numpy as np
import tensorflow as tf
from PIL import Image, UnidentifiedImageError

# ---------------------------
# Function to load Devanagari dataset
# ---------------------------
def load_devnagari_data(folder_path):
    images = []
    labels = []

    class_names = sorted(os.listdir(folder_path))
    print(f"Found classes: {class_names}")

    for label_index, folder_name in enumerate(class_names):
        full_path = os.path.join(folder_path, folder_name)
        if not os.path.isdir(full_path):
            continue  # skip non-folder items

        file_count = 0
        for filename in os.listdir(full_path):
            img_path = os.path.join(full_path, filename)

            # Only process image files
            if not filename.lower().endswith(('.png', '.jpg', '.jpeg')):
                continue

            try:
                img = Image.open(img_path).convert('L')  # grayscale
                img = img.resize((28, 28))
                images.append(np.array(img) / 255.0)
                labels.append(label_index)
                file_count += 1
            except UnidentifiedImageError:
                print(f"Skipping corrupted image: {img_path}")
                continue
            except Exception as e:
                print(f"Error loading {img_path}: {e}")
                continue

        print(f"Loaded {file_count} images for class '{folder_name}'")

    return np.array(images), np.array(labels)

# ---------------------------
# Load training and testing data
# ---------------------------
x_train, y_train = load_devnagari_data("/content/drive/MyDrive/AI ML/DevanagariHandwrittenDigitDataset/Train")
x_test, y_test   = load_devnagari_data("/content/drive/MyDrive/AI ML/DevanagariHandwrittenDigitDataset/Test")

# ---------------------------
# Convert labels to One-Hot Encoding
# ---------------------------
y_train_hot = tf.keras.utils.to_categorical(y_train, num_classes=10)
y_test_hot  = tf.keras.utils.to_categorical(y_test, num_classes=10)

# ---------------------------
# Print summary
# ---------------------------
print(f"Task Complete: Loaded {len(x_train)} training images and {len(x_test)} test images.")
print(f"x_train shape: {x_train.shape}, y_train_hot shape: {y_train_hot.shape}")
print(f"x_test shape: {x_test.shape}, y_test_hot shape: {y_test_hot.shape}")

# ---------------------------
# Optional: Visualize a few images
# ---------------------------
import matplotlib.pyplot as plt

plt.figure(figsize=(10,4))
for i in range(10):
    plt.subplot(2,5,i+1)
    plt.imshow(x_train[i], cmap='gray')
    plt.title(f"Label: {y_train[i]}")
    plt.axis("off")
plt.show()
```

Task 2: Build the FCN Model

```python
# Create Sequential Model
model = models.Sequential([
    layers.Input(shape=(28, 28)),
    layers.Flatten(),
    layers.Dense(64, activation='sigmoid'),
    layers.Dense(128, activation='sigmoid'),
    layers.Dense(256, activation='sigmoid'),
    layers.Dense(10, activation='softmax')
])
```

Task 3: Compile the Model

```python
# Compile the model
model.compile(
    optimizer='adam',                     # Optimizer
    loss='categorical_crossentropy',      # Loss function for one-hot labels
    metrics=['accuracy']                  # Metric to monitor
)

# Show model architecture
model.summary()
```

Task 4: Train the Model

```python
# Train the model
history = model.fit(
    x_train,
    y_train_hot,
    batch_size=128,
    epochs=20,
    validation_split=0.2,
    shuffle=True
)
```

Task 5: Evaluate the Model

```python
# Evaluate the model on test data
test_loss, test_acc = model.evaluate(x_test, y_test_hot, verbose=1)

# Display results
print(f"Test Accuracy: {test_acc * 100:.2f}%")
print(f"Test Loss: {test_loss:.4f}")
```

```python
# Task 6: Save the model
# ---------------------------
model.save("devnagari_digit_model.keras")
print("Model saved as devnagari_digit_model.keras")

# ---------------------------
# Task 6: Load the model
# ---------------------------
loaded_model = tf.keras.models.load_model("devnagari_digit_model.keras", compile=False)

# Re-compile for evaluation
loaded_model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

# ---------------------------
# Verification
# ---------------------------
val_loss, val_acc = loaded_model.evaluate(x_test, y_test_hot, verbose=0)
print(f"Loaded Model Verified Accuracy: {val_acc * 100:.2f}%")
```
